Step # 1: Import Libraries

```python
import pandas as pd
import sqlite3
```

Step # 2: Load Dataset & Connect To SQL Database

```python
from google.colab import files

uploaded = files.upload()


# load excel file
df = pd.read_excel("Dataset for Data Analytics (2).xlsx")

# connect SQLite in-memory database
conn = sqlite3.connect(":memory:")

# create SQL table
df.to_sql("orders", conn, index=False, if_exists="replace")
print("Table 'orders' successfully created with", len(df), "rows.")
```

Step # 3: Preview Data (Basic Select & Limit)

```python
query_1 = """
SELECT OrderID, Date, CustomerID, Product, Quantity, UnitPrice, TotalPrice
FROM orders
LIMIT 5;
"""

pd.read_sql_query(query_1, conn)
```

Step # 4: Overall Business Summary (Count, Sum, Avg)

```python
query_2 = """
SELECT
    COUNT(OrderID) AS TotalOrders,
    SUM(Quantity) AS TotalUnitsSold,
    ROUND(SUM(TotalPrice), 2) AS TotalRevenue,
    ROUND(AVG(TotalPrice), 2) AS AvgOrderValue
FROM orders;
"""

pd.read_sql_query(query_2, conn)
```

Step # 5: Filtering High-Value Orders (Where & Order By)

```python
query_3 = """
SELECT OrderID, CustomerID, Product, Quantity, TotalPrice, OrderStatus
FROM orders
WHERE TotalPrice > 2500 AND OrderStatus = 'Delivered'
ORDER BY TotalPrice DESC
LIMIT 10;
"""

pd.read_sql_query(query_3, conn)
```

Step # 6: Product Performance Analysis (Group By & Aggregations)

```python
query_4 = """
SELECT
    Product,
    COUNT(OrderID) AS TotalOrders,
    SUM(Quantity) AS TotalUnitsSold,
    ROUND(SUM(TotalPrice), 2) AS TotalRevenue,
    ROUND(AVG(TotalPrice), 2) AS AvgUnitPrice
FROM orders
GROUP BY Product
ORDER BY TotalRevenue DESC;
"""

pd.read_sql_query(query_4, conn)
```

Step # 7: Order Status Breakdown (Group By & Count)

```python
query_5 = """
SELECT
  OrderStatus,
  COUNT(OrderID) AS OrderCount,
  ROUND(SUM(TotalPrice), 2) AS TotalRevenue
FROM orders
GROUP BY OrderStatus
ORDER BY OrderCount DESC;
"""

pd.read_sql_query(query_5, conn)
```

Step # 8: Payment Method Usage & Average Spend

```python
query_6 = """
SELECT
  PaymentMethod,
  COUNT(OrderID) AS TotalTransactions,
  ROUND(AVG(TotalPrice), 2) AS AvgSpendPerOrder,
  ROUND(SUM(TotalPrice), 2) AS TotalSpend
FROM orders
GROUP BY PaymentMethod
ORDER BY TotalTransactions DESC;
"""

pd.read_sql_query(query_6, conn)
```

Step # 9: Marketing Channel / Referal Source Performance

```python
query_7 = """
SELECT
    ReferralSource,
    COUNT(OrderID) AS OrderCount,
    ROUND(SUM(TotalPrice), 2) AS TotalRevenueGenerated
FROM orders
GROUP BY ReferralSource
ORDER BY TotalRevenueGenerated DESC;
"""

pd.read_sql_query(query_7, conn)
```

Step # 10: Close Connection

```python
conn.close()
print("Database connection closed. SQL Analysis completed successfully.")
```
